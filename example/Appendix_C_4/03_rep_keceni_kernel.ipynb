```python
YGs_0 = np.zeros(n_sim); YGs_1 = np.zeros(n_sim); YGs_d = np.zeros(n_sim)
YDRs_0_exp = np.zeros(n_sim); YDRs_1_exp = np.zeros(n_sim); YDRs_d_exp = np.zeros(n_sim)
YDRs_0_box = np.zeros(n_sim); YDRs_1_box = np.zeros(n_sim); YDRs_d_box = np.zeros(n_sim)
YDRs_0_triang = np.zeros(n_sim); YDRs_1_triang = np.zeros(n_sim); YDRs_d_triang = np.zeros(n_sim)
YDRs_0_gaussian = np.zeros(n_sim); YDRs_1_gaussian = np.zeros(n_sim); YDRs_d_gaussian = np.zeros(n_sim)
```

```python
for i_sim in tqdm(range(n_sim), desc='sim', total=n_sim):
    locals().update(np.load('%s/result_%.3d.npz'%(save_dir,i_sim)))
    YGs_0[i_sim] = YG_0; YGs_1[i_sim] = YG_1; YGs_d[i_sim] = YG_d
    YDRs_0_exp[i_sim] = YDR_0_exp[id_exp]; YDRs_1_exp[i_sim] = YDR_1_exp[id_exp]; YDRs_d_exp[i_sim] = YDR_d_exp[id_exp]
    YDRs_0_box[i_sim] = YDR_0_box[id_box]; YDRs_1_box[i_sim] = YDR_1_box[id_box]; YDRs_d_box[i_sim] = YDR_d_box[id_box]
    YDRs_0_triang[i_sim] = YDR_0_triang[id_triang]; YDRs_1_triang[i_sim] = YDR_1_triang[id_triang]; YDRs_d_triang[i_sim] = YDR_d_triang[id_triang]
    YDRs_0_gaussian[i_sim] = YDR_0_gaussian[id_gaussian]; YDRs_1_gaussian[i_sim] = YDR_1_gaussian[id_gaussian]; YDRs_d_gaussian[i_sim] = YDR_d_gaussian[id_gaussian]
```
